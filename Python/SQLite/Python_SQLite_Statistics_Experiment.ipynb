```python
import sqlite3
import pandas as pd
from statistics import harmonic_mean

# Step 1: Connect to SQLite database
conn = sqlite3.connect("name_rollno.db")
cursor = conn.cursor()

# Step 2: Create table
cursor.execute("DROP TABLE IF EXISTS Name_info")

cursor.execute("""
CREATE TABLE Name_info(
    rollNo INTEGER PRIMARY KEY,
    Name VARCHAR(100),
    emailid VARCHAR(100),
    game VARCHAR(50),
    Age INTEGER
)
""")

# Step 3: Insert 10 records
records = [
    (1, "Jyoti", "jyoti@gmail.com", "Badminton", 22),
    (2, "Mona", "mona@gmail.com", "Football", 21),
    (3, "Seema", "seema@gmail.com", "Tennis", 20),
    (4, "Vivek", "vivek@gmail.com", "Cricket", 23),
    (5, "Vikash", "vikash@gmail.com", "Basketball", 24),
    (6, "Vinay", "vinay@gmail.com", "Cricket", 22),
    (7, "Puja", "puja@gmail.com", "Volleyball", 21),
    (8, "Mayank", "mayank@gmail.com", "Cricket", 25),
    (9, "Sanika", "sanika@gmail.com", "Badminton", 20),
    (10, "Laxmi", "laxmi@gmail.com", "Basketball", 22)
]

cursor.executemany("""
INSERT INTO Name_info
VALUES (?, ?, ?, ?, ?)
""", records)

# Save changes
conn.commit()

# Step 4: Read records from database
cursor.execute("SELECT * FROM Name_info")

rows = cursor.fetchall()

# Step 5: Save records into DataFrame
df = pd.DataFrame(
    rows,
    columns=["rollNo", "Name", "emailid", "game", "Age"]
)

print("Student Records:")
display(df)

# Step 6: Statistical operations on Age column
ages = df["Age"]

mean = ages.mean()
median = ages.median()
mode = ages.mode()
h_mean = harmonic_mean(ages)

print("Mean:", mean)
print("Median:", median)
print("Mode:", list(mode))
print("Harmonic Mean:", h_mean)

# Close database
conn.close()
```
