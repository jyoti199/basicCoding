```python
import csv

with open("Name_rollno.csv", "w", newline="") as file:

    fieldnames = ["Name", "emailid", "game", "Age"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        "Name": "Mayank",
        "emailid": "mayank@gmail.com",
        "game": "Cricket",
        "Age": 22
    })

    writer.writerow({
        "Name": "Sanika",
        "emailid": "sanika@gmail.com",
        "game": "Badminton",
        "Age": 21
    })

    writer.writerow({
        "Name": "mona",
        "emailid": "mona@gmail.com",
        "game": "Football",
        "Age": 22
    })

    writer.writerow({
        "Name": "Seema",
        "emailid": "seema@gmail.com",
        "game": "Tennis",
        "Age": 20
    })

    writer.writerow({
        "Name": "laxmi",
        "emailid": "laxmi@gmail.com",
        "game": "Basketball",
        "Age": 22
    })

print("CSV file created successfully.")
```
